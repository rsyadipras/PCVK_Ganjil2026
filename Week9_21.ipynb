```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

platnomor = ('/content/drive/MyDrive/PCVK2026/Images/platnomor.jpg')
```

```python
# Dilasi tanpa library

def dilasi_citra(F,w):
    #size image
    p,q= F.shape
    imgD= np.zeros((p,q), dtype=np.uint8)

    #Generate structure element dilasi (SED)
    SED= np.ones((w,w), dtype=np.uint8)
    constant1= (w-1)//2

    #Proses Dilasi
    for i in range(constant1, p-constant1):
      for j in range(constant1,q-constant1):
        temp= F[i-constant1:i+constant1+1, j-constant1:j+constant1+1]
        product= temp*SED
        imgD[i,j]= np.max(product)
    return imgD

img_d= cv2.imread(platnomor,0)
ret, thresh = cv2.threshold(img_d, 125, 200, cv2.THRESH_BINARY)

w=5
imgD = dilasi_citra(thresh,w)
plt.subplot(131),plt.imshow(img_d,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
plt.subplot(132),plt.imshow(imgD,cmap = 'gray')
plt.title('Citra Hasil Dilasi'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
# Erosi Tanpa Library

def erosi_citra(F,k):
  #size image
  m,n= F.shape

  #Generate structure element erosi (SE)
  SE= np.ones((k,k), dtype=np.uint8)
  constant= (k-1)//2
  imgE= np.zeros((m,n), dtype=np.uint8)

  #Proses Erosi
  for i in range(constant, m-constant):
    for j in range(constant,n-constant):
      temp= F[i-constant:i+constant+1, j-constant:j+constant+1]
    product= temp*SE
    imgE[i,j]= np.min(product)
  return imgE

img_e= cv2.imread(platnomor,0)
ret1, thresh1 = cv2.threshold(img_e, 175, 225, cv2.THRESH_BINARY)
k=5
imgE = erosi_citra(thresh1,k)

plt.subplot(131),plt.imshow(img_e,cmap = 'gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
plt.subplot(132),plt.imshow(imgE,cmap = 'gray')
plt.title('Citra Hasil Erosi'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
import pandas as pd
import pytesseract
import difflib

# Setup Tesseract & Config
LANG = 'ind' if 'ind' in pytesseract.get_languages() else 'eng'
CONFIG = '--psm 6'

# Load Citra KTP
PATH_KTP = '/content/drive/MyDrive/PCVK2026/Images/ktp.jpeg'
img_bgr = cv.imread(PATH_KTP)
img_bgr = cv.resize(img_bgr, (1000, 630))
gray = cv.cvtColor(img_bgr, cv.COLOR_BGR2GRAY)
blur = cv.GaussianBlur(gray, (5, 5), 0)

# Penerapan Metode Morfologi Citra KTP

# Baseline: Otsu Thresholding saja
_, v1_otsu = cv.threshold(blur, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)

# Metode A: Adaptive Thresholding + Opening (Membuang Noise Bintik Kecil)
adp = cv.adaptiveThreshold(
    cv.medianBlur(gray, 3), 255,
    cv.ADAPTIVE_THRESH_GAUSSIAN_C, cv.THRESH_BINARY_INV, 31, 10
)
se_open = cv.getStructuringElement(cv.MORPH_RECT, (2, 2))
adp_open = cv.morphologyEx(adp, cv.MORPH_OPEN, se_open)
v2_adaptive_open = cv.bitwise_not(adp_open) # Inversi teks hitam latar putih

# Metode B: Black-Hat + Otsu + Closing (Menonjolkan Teks & Menyambung Huruf Putus)
se_blackhat = cv.getStructuringElement(cv.MORPH_RECT, (13, 5))
blackhat = cv.morphologyEx(blur, cv.MORPH_BLACKHAT, se_blackhat)
_, bh_bin = cv.threshold(blackhat, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)

se_close = cv.getStructuringElement(cv.MORPH_RECT, (2, 2))
bh_close = cv.morphologyEx(bh_bin, cv.MORPH_CLOSE, se_close)
v3_blackhat_close = cv.bitwise_not(bh_close) # Inversi teks hitam latar putih

# Visualisasi Perbandingan Citra Hasil Morfologi
masukan = {
    '1. Otsu (Baseline)': v1_otsu,
    '2. Adaptive + Opening': v2_adaptive_open,
    '3. Black-Hat + Otsu + Closing': v3_blackhat_close
}

plt.figure(figsize=(18, 5))
plt.subplot(1, 4, 1)
plt.imshow(gray, cmap='gray')
plt.title('Citra Keabuan Asli')
plt.axis('off')

for i, (nama, im_) in enumerate(masukan.items(), 2):
    plt.subplot(1, 4, i)
    plt.imshow(im_, cmap='gray')
    plt.title(nama)
    plt.axis('off')

plt.tight_layout()
plt.show()

# Pembacaan OCR & Evaluasi Ketepatan
semua_masukan = {'Asli (Gray)': gray}
semua_masukan.update(masukan)

teks_ocr, conf_semua = {}, {}

for nama, im_ in semua_masukan.items():
    teks_ocr[nama] = pytesseract.image_to_string(im_, lang=LANG, config=CONFIG)
    d = pytesseract.image_to_data(im_, lang=LANG, config=CONFIG, output_type=pytesseract.Output.DICT)

    conf_list = [float(d['conf'][j]) for j in range(len(d['text'])) if float(d['conf'][j]) >= 0]
    conf_semua[nama] = conf_list

# Daftar 10 Kata Acuan pada KTP
ACUAN = [
    'PROVINSI', 'KOTA', 'NIK', 'PANDANWANGI',
    'TEMPAT', 'MALANG', 'ADI', 'RT/RW',
    'ISLAM', 'JAWA'
]

def token_ocr(teks):
    return [t.strip() for t in teks.split() if t.strip()]

benar = {n: 0 for n in semua_masukan}
for a in ACUAN:
    for n in semua_masukan:
        cocok = any(t.lower() == a.lower() for t in token_ocr(teks_ocr[n]))
        benar[n] += int(cocok)

# Ringkasan Evaluasi
ringkasan = pd.DataFrame({
    'Masukan OCR': list(semua_masukan),
    'Item Terbaca Tepat': [benar[n] for n in semua_masukan],
    'Kecocokan (%)': [benar[n] / len(ACUAN) * 100 for n in semua_masukan],
    'Rata-rata Confidence': [
        round(float(np.mean(conf_semua[n])), 1) if conf_semua[n] else 0
        for n in semua_masukan
    ]
})

display(ringkasan)
```

```python
# Tugas Praktikum - Analisis

# 1. Membaca citra KTM dalam format RGB
ktm = io.imread('/content/drive/MyDrive/PCVK2026/KTM_01.jpg')

# 2. Menhitung rata-rata intensitas RGB tiap piksel
avg_pix = np.mean(ktm, axis=2)

# 3. Membuat boolean mask untuk piksel terang (rata-rata > 128)
pix_terang = avg_pix > 128

# 4. Menghitung persentase piksel terang
total_pixels = ktm.shape[0] * ktm.shape[1]
jml_pixterang = np.sum(pix_terang)
persentase = (jml_pixterang / total_pixels) * 100

# 5. Menghitung rata-rata intensitas kanal R, G, B keseluruhan citra
mean_allcitra = np.mean(ktm, axis=(0, 1))

# 6. Menghitung rata-rata intensitas kanal R, G, B khusus area terang
bright_pixels = ktm[pix_terang]
mean_pixterang = np.mean(bright_pixels, axis=0)

# Output Data
print(f"Total Piksel Citra          : {total_pixels} piksel")
print(f"Jumlah Piksel Terang (>128) : {jml_pixterang} piksel")
print(f"Persentase Piksel Terang    : {persentase:.2f}%\n")
print(f"Rata-rata RGB Keseluruhan   : R={mean_allcitra[0]:.2f}, G={mean_allcitra[1]:.2f}, B={mean_allcitra[2]:.2f}")
print(f"Rata-rata RGB Area Terang   : R={mean_pixterang[0]:.2f}, G={mean_pixterang[1]:.2f}, B={mean_pixterang[2]:.2f}")
```

```python
# Tugas Praktikum - Aplikasi No 01

def tambahkan_label_identitas(img, teks="Dokumentasi PCVK - 2026"):
    img_out = img.copy()
    h, w = img_out.shape[:2]

    # Menghitung tinggi bar adaptif (h // 12)
    bar_h = h // 12

    # Menggambar bar hitam di bagian bawah
    cv.rectangle(img_out, (0, h - bar_h), (w, h), (0, 0, 0), -1)
    font = cv.FONT_HERSHEY_SIMPLEX

    # Menghitung skala font awal berbasis tinggi bar
    font_scale = bar_h / 40.0
    thickness = max(1, int(font_scale * 1.5))

    # Hitung ukuran piksel teks awal
    (text_w, text_h), _ = cv.getTextSize(teks, font, font_scale, thickness)

    # Penyesuaian adaptif lebar, jika teks melebihi 90% lebar citra, kecilkan font_scale
    max_teks_w = int(w * 0.90)
    if text_w > max_teks_w:
        font_scale = font_scale * (max_teks_w / text_w)
        thickness = max(1, int(font_scale * 1.5))
        (text_w, text_h), _ = cv.getTextSize(teks, font, font_scale, thickness)

    # menentukan posisi teks tepat di tengah bar
    text_x = (w - text_w) // 2
    text_y = h - (bar_h - text_h) // 2

    # membubuhkan teks putih
    cv.putText(img_out, teks, (text_x, text_y), font, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return img_out, bar_h

# membaca citra asli
img_lanskap = io.imread('/content/drive/MyDrive/PCVK2026/gmbr1.jpg')
img_potret = io.imread('/content/drive/MyDrive/PCVK2026/gmbr2.jpg')

# proses citra
res_lanskap, bar_l = tambahkan_label_identitas(img_lanskap)
res_potret, bar_p = tambahkan_label_identitas(img_potret)

# visualisasi
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Baris 1: Citra Lanskap
axes[0, 0].imshow(img_lanskap)
axes[0, 0].set_title(f"Masukan 1 (Lanskap): {img_lanskap.shape[1]} x {img_lanskap.shape[0]} px")
axes[0, 0].axis('off')

axes[0, 1].imshow(res_lanskap)
axes[0, 1].set_title(f"Keluaran: bar = {img_lanskap.shape[0]}//12 = {bar_l} px")
axes[0, 1].axis('off')

# Baris 2: Citra Potret
axes[1, 0].imshow(img_potret)
axes[1, 0].set_title(f"Masukan 2 (Potret): {img_potret.shape[1]} x {img_potret.shape[0]} px")
axes[1, 0].axis('off')

axes[1, 1].imshow(res_potret)
axes[1, 1].set_title(f"Keluaran: bar = {img_potret.shape[0]}//12 = {bar_p} px")
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Tugas Praktikum - Aplikasi No 02

def buat_thumbnail_marketplace(img, nama_toko="Adi's Store", warna_kanvas=(240, 240, 240)):
    img_out = img.copy()
    h, w = img_out.shape[:2]

    # Menentukan dimensi kanvas persegi (1:1) dari nilai maksimum dimensi citra
    canvas_size = max(h, w)

    # Membuat kanvas persegi dengan warna latar belakang netral
    kanvas = np.full((canvas_size, canvas_size, 3), warna_kanvas, dtype=np.uint8)

    # Menghitung offset untuk menempatkan citra tepat di tengah kanvas
    x_offset = (canvas_size - w) // 2
    y_offset = (canvas_size - h) // 2

    # Menempelkan citra masukan pada kanvas menggunakan pengirisan NumPy
    kanvas[y_offset:y_offset + h, x_offset:x_offset + w] = img_out

    # Menentukan dimensi badge logo toko secara proporsional terhadap ukuran kanvas
    badge_w = int(canvas_size * 0.32)
    badge_h = int(canvas_size * 0.08)
    margin = int(canvas_size * 0.03)

    # Koordinat badge di sudut kanan atas
    pt1 = (canvas_size - badge_w - margin, margin)
    pt2 = (canvas_size - margin, margin + badge_h)

    # Menggambar latar belakang badge
    cv.rectangle(kanvas, pt1, pt2, (35, 35, 35), -1)

    # Menghitung skala font dan ketebalan secara adaptif terhadap badge
    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = badge_h / 40.0
    thickness = max(1, int(font_scale * 1.5))

    # Penyesuaian otomatis agar teks tidak melampaui lebar badge
    (text_w, text_h), _ = cv.getTextSize(nama_toko, font, font_scale, thickness)
    max_text_w = int(badge_w * 0.85)
    if text_w > max_text_w:
        font_scale = font_scale * (max_text_w / text_w)
        thickness = max(1, int(font_scale * 1.5))
        (text_w, text_h), _ = cv.getTextSize(nama_toko, font, font_scale, thickness)

    # Menentukan posisi teks tepat di tengah badge
    text_x = pt1[0] + (badge_w - text_w) // 2
    text_y = pt1[1] + (badge_h + text_h) // 2

    # Membubuhkan teks nama toko pada badge
    cv.putText(kanvas, nama_toko, (text_x, text_y), font, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return kanvas

# Read input citra
img_lanskap = io.imread('/content/drive/MyDrive/PCVK2026/gmbr1.jpg')
img_potret = io.imread('/content/drive/MyDrive/PCVK2026/gmbr2.jpg')

# processing citra
thumb_lanskap = buat_thumbnail_marketplace(img_lanskap, "Adi's Store")
thumb_potret = buat_thumbnail_marketplace(img_potret, "Adi's Store")

# visualize
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Thumbnail 1: Hasil Landscape
axes[0].imshow(thumb_lanskap)
axes[0].set_title(f"Thumbnail Landscape (1:1)\n{thumb_lanskap.shape[1]} x {thumb_lanskap.shape[0]} px")
axes[0].axis('off')

# Thumbnail 2: Hasil Potret
axes[1].imshow(thumb_potret)
axes[1].set_title(f"Thumbnail Potret (1:1)\n{thumb_potret.shape[1]} x {thumb_potret.shape[0]} px")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Untuk menampilkan Gambar
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
import matplotlib.pyplot as plt
import numpy as np

img = cv.imread('ktp.jpeg')
cv2_imshow(img)

print("Resolusi Gambar: ", img.shape[1], " x ", img.shape[0])
```

```python
# Menampilkan gambar

img = cv.imread('ktp.jpeg')
plt.imshow(img) # Tampilkan Format BGR
```

```python
img = cv.imread('ktp.jpeg')
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB)) # Tampilkan Format BGR -> RGB
plt.show()
```

```python
# Menampilkan Citra Grayscale
img_gray = cv.imread('ktp.jpeg', cv.IMREAD_GRAYSCALE)

# Tanpa CMAP
plt.subplot(1,2,1)
plt.imshow(img_gray)
plt.title('Tanpa CMAP')

# CMAP
plt.subplot(1,2,2)
plt.imshow(img_gray, cmap='gray')
plt.title('CMAP')

plt.show()
```

```python
# Melakukan Resizing
imgRGB = cv.cvtColor(img, cv.COLOR_BGR2RGB) #konversi BGR ke RGB
img_output = cv.resize(imgRGB, (347, 229))
plt.imshow(img_output)
```

```python
# Melakukan Flipping
imgFlip = cv.flip(img_output, 1)

# tampil ukuran canvas lebih besar
fig = plt.figure(figsize=(10,10))
axz = fig.add_subplot(1,1,1)
plt.imshow(imgFlip)
```

```python
# Membuat bentuk Geometri 2D dari OpenCV
black_img = np.zeros(shape=(530, 530, 3), dtype=np.int16)

# menambahkan bentuk persegi koordinat pt1 dan pt2
cv.rectangle(black_img, pt1=(290, 0), pt2=(520,130), color=(0,0,255), thickness=10)

cv.rectangle(black_img, pt1=(250, 250), pt2=(350,350), color=(255,0,0), thickness=15)

cv.circle(black_img, center=(150, 150), radius=60, color=(0, 255, 0), thickness=8)

cv.circle(black_img, center=(400, 400), radius=50, color=(0, 255, 0), thickness=-1)
cv.line(black_img, pt1=(0,0), pt2=(530,530), color=(255,255,255), thickness=5)

font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img, text='PCVK is Fun!' , org=(10, 500), fontFace=font,
           fontScale=2, color=(255,255,0), thickness=3, lineType=cv.LINE_AA)

# tampilkan gambar
plt.imshow(black_img)
```

```python
# Pertanyaan Praktikum No. 04
# 1. Membaca citra dengan skimage
img_skimage = io.imread('/content/drive/MyDrive/PCVK2026/KTM_01.jpg')

# 2. Menerapkan cv2.cvtColor pada citra skimage
img_cvt = cv.cvtColor(img_skimage, cv.COLOR_BGR2RGB)

# 3. Mengambil nilai piksel pada koordinat yang sama (y=50, x=50)
y, x = 50, 50
p_ril = img_skimage[y, x]
p_tukar = img_cvt[y, x]

# 4. Menampilkan gambar secara berdampingan
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_skimage)
plt.title("skimage.io.imread()")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_cvt)
plt.title("cv2.cvtColor()")
plt.axis('off')

plt.show()

print(f"Piksel skimage.io.imread() : {p_ril}")
print(f"Piksel cv2.cvtColor()      : {p_tukar}")
```

```python
# Tugas Praktikum Penerapan 01
# 1. Membaca citra dalam format RGB
img = io.imread('/content/drive/MyDrive/PCVK2026/KTM_01.jpg')

# 2. Kombinasi Merah-Biru (Kanal Hijau pada indeks 1 dinolkan)
img_mb = img.copy()
img_mb[:, :, 1] = 0

# 3. Kombinasi Hijau-Biru (Kanal Merah pada indeks 0 dinolkan)
img_hb = img.copy()
img_hb[:, :, 0] = 0

# 4. Menampilkan kedua hasil berdampingan
plt.figure(figsize=(10, 5))
plt.suptitle("Perbandingan Kombinasi Kanal Warna", fontsize=14)

plt.subplot(1, 2, 1)
plt.imshow(img_mb)
plt.title("Kombinasi Merah-Biru\n(Kanal Hijau = 0)")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_hb)
plt.title("Kombinasi Hijau-Biru\n(Kanal Merah = 0)")
plt.axis('off')

plt.show()
```

```python
# Tugas Praktikum Penerapan 02
# 1. Membaca citra dalam format RGB
img_ktm = io.imread('/content/drive/MyDrive/PCVK2026/KTM_01.jpg')

# 2. Menampilkan ketiga hasil secara berdampingan
plt.figure(figsize=(10, 3))
plt.suptitle("3 Bentuk Pencerminan", fontsize=14)

plt.subplot(1, 3, 1)
plt.imshow(cv.flip(img_ktm, 0)) # vertikal bernilai 0
plt.title("Cermin Vertikal")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv.flip(img_ktm, 1)) # horizontal bernilai 1
plt.title("Cermin Horizontal")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv.flip(img_ktm, -1)) # keduanya bernilai -1
plt.title("Cermin Keduanya")
plt.axis('off')

plt.show()
```
